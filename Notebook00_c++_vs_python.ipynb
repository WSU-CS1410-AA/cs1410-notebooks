# C++ vs Python

Here is an example program that performs a computationally intensive operation hundreds of millions of times to illustrate how fast C++ programs are compared to languages like Python.



## First: Using pure Python

Here is the Python version:

```python
import time
from math import log

N = int(1e8)
r = 0

start = time.time()

for i in range(1, N + 1):
    r = i * i + log(i)

end = time.time()

print("Last square and log:", r)
print("Time taken:", end - start, "seconds")
```

## Second: Using C++

Here is the same program in C++.

```python
%%writefile main.cpp

#include <iostream>
#include <cmath>
#include <chrono>

int main() {
    const long long N = 1e8;
    long long r = 0;

    auto start = std::chrono::high_resolution_clock::now();

    for (long long i = 1; i <= N; ++i) {
        r = i * i + std::log(i);
    }

    auto end = std::chrono::high_resolution_clock::now();
    std::chrono::duration<double> elapsed = end - start;

    std::cout << "Last square and log: " << r << std::endl;
    std::cout << "Time taken: " << elapsed.count() << " seconds\n";

    return 0;
}
```

```python
!g++ -O2 main.cpp -o main
!./main
```

Wow! That’s a significant difference. C++ is over 40 times faster than Python. This is why, more than fifty years later, C and C++ remain as relevant as ever.

## Third: Using Python with NumPy
In Python, popular packages like NumPy are used for intensive computation. Here’s another Python program that does the same.

```python
import numpy as np
import time

N = int(1e8)

start = time.time()

arr = np.arange(1, N + 1, dtype=np.int64)
squares_and_logs = arr * arr + np.log(arr)

end = time.time()

print("Last square and log:", squares_and_logs[-1])
print("Time taken:", end - start, "seconds")
```

That is much better. C++ is only about 2 times faster. But, why is there such a significant difference between these two Python versions? Well! The NumPy version is faster than the pure Python version because NumPy is actually implemented in C and C++, which emphasizes the importance of C/C++ in implementing efficient important programs and libraries.
